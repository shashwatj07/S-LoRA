```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rps_list = [10, 14]
duration = 600
```

```python
for rps in rps_list:
    prod_trace = pd.read_csv(f"prod_{rps}.0_600_50_adapters.csv")
    original_trace = pd.read_csv(f"../experiment_traces_v2/uniform_poisson_{rps}.0_600.csv")
    
    # keep the req_time from the original_trace and the other columns from the prod_trace
    # take min of rows of both dataframes
    
    min_rows = min(len(prod_trace), len(original_trace))
    combined_trace = pd.DataFrame()
    combined_trace['req_time'] = original_trace['req_time'][:min_rows]
    combined_trace['model_dir'] = prod_trace['model_dir'][:min_rows]
    combined_trace['adapter_dir'] = prod_trace['adapter_dir'][:min_rows]
    combined_trace['prompt_len'] = prod_trace['prompt_len'][:min_rows]
    combined_trace['output_len'] = prod_trace['output_len'][:min_rows]
    combined_trace['req_id'] = prod_trace['req_id'][:min_rows]
    
    # save to csv with column order req_id,model_dir,adapter_dir,prompt_len,output_len,req_time
    
    combined_trace = combined_trace[['req_id', 'model_dir', 'adapter_dir', 'prompt_len', 'output_len', 'req_time']]
    combined_trace.to_csv(f"prod_combined_{rps}_600_50_adapters.csv", index=False)
```
